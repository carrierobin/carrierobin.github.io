```python
import requests
import json
import pandas as pd
```

```python
from dotenv import load_dotenv
import os

load_dotenv()  # loads variables from .env into environment

brightdata_api_key = os.getenv('BRIGHTDATA_API_KEY')
```

```python
# Load environment variables from .env file
load_dotenv()

# Get API key from environment
BRIGHTDATA_API_KEY = os.getenv("BRIGHTDATA_API_KEY")

def trigger_dataset(api_token, dataset_id, company_url, csv_filename="output.csv"):
    """
    Triggers a dataset using the BrightData API and saves the result to a CSV.

    Args:
    api_token (str): The API token for authentication.
    dataset_id (str): The dataset ID to trigger.
    company_url (str): The URL of the company page to analyze.
    csv_filename (str): The filename for saving CSV output.

    Returns:
    dict: The JSON response from the API.
    """
    headers = {
        "Authorization": f"Bearer {api_token}",
        "Content-Type": "application/json",
    }
    payload = json.dumps([{"url": company_url}])
    response = requests.post(
        "https://api.brightdata.com/datasets/v3/trigger",
        headers=headers,
        params={"dataset_id": dataset_id},
        data=payload,
    )

    data = response.json()

    # Convert to CSV
    if isinstance(data, list):
        df = pd.DataFrame(data)
    elif 'result' in data:
        df = pd.DataFrame(data['result'])
    else:
        df = pd.DataFrame([data])

    df.to_csv(csv_filename, index=False)
    print(f"CSV saved as '{csv_filename}'")

    return data

# Example usage
dataset_id = "gd_l7j1po0921hbu0ri1z"
company_url = "https://www.glassdoor.com/Overview/Working-at-DaVita-EI_IE1432.11,17.htm"

response_data = trigger_dataset(BRIGHTDATA_API_KEY, dataset_id, company_url, csv_filename="davita_reviews.csv")
```
